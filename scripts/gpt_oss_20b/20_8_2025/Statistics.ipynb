```python
from dotenv import load_dotenv
import os
load_dotenv()
file_review = os.getenv("File_Output")
print(file_review)
```

## Thống kê F1-Score

```python
import pandas as pd
df = pd.read_csv(file_review)
df.info()
```

```python
import pandas as pd

num_samples = len(df)

precision_sum = df["precision"].sum()
recall_sum = df["recall"].sum()
f1_sum = df["f1_score"].sum()

precision_avg = df["precision"].mean()
recall_avg = df["recall"].mean()
f1_avg = df["f1_score"].mean()

print(f"Tổng số mẫu: {num_samples}")
print(f"Tổng Precision: {precision_sum:.2f}/{num_samples} (Trung bình: {precision_avg:.2f})")
print(f"Tổng Recall: {recall_sum:.2f}/{num_samples} (Trung bình: {recall_avg:.2f})")
print(f"Tổng F1-score: {f1_sum:.2f}/{num_samples} (Trung bình: {f1_avg:.2f})")

#Các mẫu fail (fail = điểm < 1)
fail_precision = df[df["precision"] < 1]
fail_recall = df[df["recall"] < 1]
fail_f1 = df[df["f1_score"] < 1]

print(f"\nSố mẫu Precision fail: {len(fail_precision)}")
print(f"row_index fail Precision: {fail_precision['row_index'].tolist()}")

print(f"\nSố mẫu Recall fail: {len(fail_recall)}")
print(f"row_index fail Recall: {fail_recall['row_index'].tolist()}")

print(f"\nSố mẫu F1 fail: {len(fail_f1)}")
print(f"row_index fail F1: {fail_f1['row_index'].tolist()}")
```

## Thống kê Groundedness

```python
df2 = pd.read_csv(file_review)
df2.info()
```

```python
import pandas as pd


# 1. Thống kê số lượng sample
num_samples = len(df2)

# 2. Tổng và trung bình groundedness
groundedness_sum = df2["groundedness_score"].sum()
groundedness_avg = df2["groundedness_score"].mean()

print(f"Tổng số mẫu: {num_samples}")
print(f"Tổng Groundedness Score: {groundedness_sum:.2f} (Trung bình: {groundedness_avg:.2f})")

# 3. Liệt kê các mẫu fail
fail_groundedness = df2[df2["groundedness_score"] < 1]

print(f"\nSố mẫu Groundedness fail: {len(fail_groundedness)}")
print(f"row_index fail Groundedness: {fail_groundedness['row_index'].tolist()}")
```

## Thống kê noisesensitivy

```python
import pandas as pd
df3 = pd.read_csv(file_review)
df3.info()
```

```python
import pandas as pd

num_samples = len(df3)
noise_sum = df3["noisesensitivy_score"].sum()
noise_avg = df3["noisesensitivy_score"].mean()

print(f"Tổng số mẫu: {num_samples}")
print(f"Tổng Noise Sensitivity Score: {noise_sum:.2f} (Trung bình: {noise_avg:.2f})")

# mẫu fail > 0 
fail_noise = df3[df3["noisesensitivy_score"] >0]

print(f"\nSố mẫu Noise Sensitivity fail: {len(fail_noise)}")
print(f"row_index fail Noise Sensitivity: {fail_noise['row_index'].tolist()}")
```
